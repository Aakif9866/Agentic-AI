# Python Core Basics — Recap

Not a full Python course. Just the core syntax every later module assumes you already read comfortably: variables, control flow, collections, functions, classes, imports. If every cell here is already obvious, skip straight to Module 1.

## 1. Variables & basic types

```python
name = "Aakif"             # str
age = 25                   # int
height = 5.9               # float
is_learning = True         # bool
nothing_yet = None         # None = "no value"

print(type(name), type(age), type(height), type(is_learning), type(nothing_yet))
```

## 2. Operators

```python
a, b = 7, 2
print(a + b, a - b, a * b, a / b, a // b, a % b, a ** b)   # arithmetic
print(a > b, a == b, a != b)                                # comparison
print(a > 0 and b > 0, a > 0 or b < 0, not is_learning)     # logical
```

## 3. Strings

```python
greeting = f"Hello, {name}! You are {age} years old."
print(greeting)
print(name.upper(), name.lower(), len(name))
print(name[0], name[-1], name[0:3])   # indexing & slicing
```

## 4. Collections: list, tuple, dict, set

```python
modules = ["foundations", "python", "langchain"]   # list: ordered, mutable
modules.append("langgraph")
print(modules, modules[0], len(modules))

coordinates = (28.6, 77.2)                          # tuple: ordered, immutable
print(coordinates[0])

student = {"name": "Aakif", "level": "beginner"}    # dict: key -> value
student["goal"] = "agentic AI"
print(student["name"], student.get("missing_key", "default"))

unique_tags = {"ai", "python", "ai"}                # set: duplicates collapse
print(unique_tags)
```

## 5. Control flow: if / elif / else

```python
score = 72

if score >= 90:
    grade = "A"
elif score >= 70:
    grade = "B"
else:
    grade = "C"

print(f"Score {score} -> grade {grade}")
```

## 6. Loops: for, while, range, enumerate

```python
for module in modules:
    print("studying:", module)

for i, module in enumerate(modules):
    print(i, module)

count = 0
while count < 3:
    print("count:", count)
    count += 1
```

## 7. Functions: def, defaults, *args/**kwargs, return

```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Aakif"))
print(greet("Aakif", greeting="Hi"))

def total(*numbers, **labels):
    print("labels:", labels)
    return sum(numbers)

print(total(1, 2, 3, source="module0"))
```

## 8. Classes: just enough to recognise them later

```python
class Module:
    def __init__(self, number, title):
        self.number = number
        self.title = title

    def describe(self):
        return f"Module {self.number}: {self.title}"

m = Module(0, "Python Core Basics")
print(m.describe())
```

`self` is the instance itself. `__init__` runs once, when the object is created. This is as far as classes go here — Module 2's Pydantic `BaseModel` builds on this exact same `class` syntax.

## 9. Imports & modules

```python
import math
from math import sqrt

print(math.pi, sqrt(16))

if __name__ == "__main__":
    print("This only runs when the file is executed directly, not when imported.")
```
